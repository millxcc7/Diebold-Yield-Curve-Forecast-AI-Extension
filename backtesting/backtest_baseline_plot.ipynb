```python

import pandas as pd 
import plotly.graph_objects as go


# All backtest scenarios (folder names inside results/)
series = ["covid_2020", "economic_recovery_2014", "financial_crisis_2008", "holdout_2025",
          "low_volatility_2017", "normal_policy_adjust_2019", "rate_hikes_2022"]

BABY_BLUE = "#89CFF0"
BABY_PINK = "#F4C2C2"

for s in series:
    # Load this scenario's backtest results (rows = month-end dates, columns = maturities)
    actual = pd.read_csv(f"results/{s}/baseline_actual_yields.csv", index_col=0, parse_dates=True)
    predicted = pd.read_csv(f"results/{s}/baseline_predicted_yields.csv", index_col=0, parse_dates=True)

    maturities = [int(c) for c in actual.columns]   # 3, 6, ..., 120 months

    fig = go.Figure()

    # Actual yields (baby blue)
    fig.add_trace(go.Surface(
        x=maturities, y=actual.index, z=actual.values,
        colorscale=[[0, BABY_BLUE], [1, BABY_BLUE]],   # one solid color
        showscale=False, opacity=0.9,
        name="Actual", showlegend=True,
    ))

    # Predicted yields (baby pink)
    fig.add_trace(go.Surface(
        x=maturities, y=predicted.index, z=predicted.values,
        colorscale=[[0, BABY_PINK], [1, BABY_PINK]],
        showscale=False, opacity=0.9,
        name="Predicted (Diebold-Li AR(1))", showlegend=True,
    ))

    fig.update_layout(
        title=f"{s.replace('_', ' ').title()}: Actual vs Predicted Yield Curve",   # e.g. "Covid 2020"
        scene=dict(
            xaxis_title="Maturity (months)",
            yaxis_title="Date",
            zaxis_title="Yield (%)",
        ),
        width=900, height=700,
    )
    fig.show()
```
